```python
# Boxplots de las 13 variables del conjunto Wine

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_wine

# 1. Cargar y organizar los datos
wine = load_wine()

df_box = pd.DataFrame(
    wine.data,
    columns=wine.feature_names
)

df_box["clase"] = [
    f"class_{i}" for i in wine.target
]

# 2. Definir clases, colores y variables
clases = ["class_0", "class_1", "class_2"]

colores = {
    "class_0": "#4C78A8",
    "class_1": "#F58518",
    "class_2": "#54A24B"
}

variables = wine.feature_names

# 3. Crear una cuadrícula de 4 filas y 4 columnas
fig, axes = plt.subplots(4, 4, figsize=(20, 15))
axes = axes.flatten()

# 4. Crear un boxplot para cada variable
for ax, variable in zip(axes, variables):

    datos = [
        df_box.loc[
            df_box["clase"] == clase, variable
        ].values
        for clase in clases
    ]

    cajas = ax.boxplot(
        datos,
        positions=[1, 2, 3],
        patch_artist=True,
        widths=0.6,
        showfliers=True
    )

    # Asignar un color a cada clase
    for caja, clase in zip(cajas["boxes"], clases):
        caja.set_facecolor(colores[clase])
        caja.set_alpha(0.85)

    ax.set_title(
        variable.replace("_", " ").title(),
        fontsize=10
    )

    ax.set_xticks([1, 2, 3])
    ax.set_xticklabels(clases, fontsize=8)
    ax.set_ylabel("")
    ax.grid(axis="y", linestyle="--", alpha=0.3)

# 5. Ocultar los tres espacios que sobran
for ax in axes[len(variables):]:
    ax.set_visible(False)

# 6. Título general y presentación
fig.suptitle(
    "Distribución de las variables químicas por clase de vino",
    fontsize=16
)

plt.tight_layout(rect=[0, 0, 1, 0.96])

# Guardar la imagen en alta resolución
plt.savefig(
    "boxplots_wine.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
